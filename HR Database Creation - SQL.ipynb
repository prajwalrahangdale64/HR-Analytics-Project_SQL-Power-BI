# <u>Creating Database named</u> hr_analytics_project

- ***STEP 1:*** Right-click postgres database → Query Tool

```sql

CREATE DATABASE hr_analytics_project;

```

---

# <u>Importing data from the</u> Human Resources.csv <u>file.

- ***STEP 1:*** Right-click hr_analytics_project database → Query Tool

- ***STEP 2:*** Created an empty table (hr) with the cloumns that match up the Human Resources.csv file columns.

```sql

CREATE TABLE hr(
    "id" VARCHAR(20),
    "first_name" VARCHAR(100),
    "last_name" VARCHAR(100),
    "birthdate" VARCHAR(20),
    "gender" VARCHAR(20),
    "race" VARCHAR(100),
    "department" VARCHAR(100),
    "jobtitle" VARCHAR(100),
    "location" VARCHAR(50),
    "hire_date" VARCHAR(20),
    "termdate" VARCHAR(50),
    "location_city" VARCHAR(50),
    "location_state" VARCHAR(50)
);

```

- ***STEP 3:*** Expand hr_analytics_project → Schemas → public → Tables → Right-click hr → Import/Export Data

    

    - Import Human Resources.csv file, turn Header = ON (because CSV has column names). Click OK.

    

    

- ***STEP 4:*** Checking

```sql

SELECT * FROM hr;

```
